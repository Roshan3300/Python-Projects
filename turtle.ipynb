```python
import turtle, colorsys
t = turtle.Turtle()
t.speed(0)
turtle.bgcolor("black")
turtle.colormode(255)
for i in range(150):
    c= colorsys.hsv_to_rgb(i/100,1,1)
    t.pencolor(int(c[0]*255), int(c[1]*255), int(c[2]*255))
    for _ in range(4):
        t.forward(i*3)
        t.right(90)
    t.right(15)

turtle.done()
```

```python
import turtle
t=turtle.Turtle(); t.speed(0); turtle.bgcolor("black");
t.color("cyan")

def sierpinski(length, depth):
    if depth==0:
        for _ in range(3): t.forward(length); t.left(120)
    else:
        sierpinski(length/2, depth -1)
        t.forward(length/2)
        sierpinski(length/2, depth -1)
        t.backward(length/2); t.left(60)
        t.forward(length/2); t.right(60)
        sierpinski(length/2, depth -1)
        t.left(60); t.backward(length/2); t.right(60)
sierpinski(300, 4)
turtle.done()
```

```python
import numpy as np
import pandas as pd
```

```python
foods=["milk","biskut","beread","tost","appel"]
days=[f"Day{i}" for i in range(1,8)]
```

```python
calories=np.array([
    [100,200,300,400,500],
    [100,120,130,140,150],
    [200,210,220,230,240],
    [300,310,320,330,340],
    [400,410,420,430,440],
    [500,510,520,530,540],
    [600,610,620,630,640]
])
calories
```

```python
total_calories_day=np.sum(calories, axis=1)
print("total calories= ")
for day, total in zip(days, total_calories_day):
    print(f"{day}: {total} calories")
```

```python
avg_calories_food=np.mean(calories,axis=0)
print("average calorie food= ")
for food, avg in zip(foods, avg_calories_food):
    print(f"{food}: {avg:.2f} calories" )
```

```python
highest_day_index=np.argmax(total_calories_day)
highest_day=days[highest_day_index]
print(f"the calories intake= {highest_day}"
      f"with{total_calories_day[highest_day_index]}kcal.")
```

```python
df=pd.DataFrame(calories, columns=foods, index=days)
print("calories=")
display(df)
```
