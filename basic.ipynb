```python
# decordar
def my_decorator(func):
    def wrapper():
        print("👋 Before the function runs")
        func()
        print("👋 After the function runs")
    return wrapper

@my_decorator
def greet():
    print("Hello, Rashmi!")

greet()


def decorder(func):
    def wrapper():
        print('Before the fuction runs')
        func()
        print('After the function run')
    return wrapper

@decorder
def greet():
    print('hello alll')

greet()
```
